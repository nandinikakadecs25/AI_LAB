```python
def pl_true(sentence, model):
    if isinstance(sentence, str):
        return model.get(sentence, False)

    operator = sentence[0]
    if operator == 'not':
        return not pl_true(sentence[1], model)
    elif operator == 'and':
        return all(pl_true(arg, model) for arg in sentence[1:])
    elif operator == 'or':
        return any(pl_true(arg, model) for arg in sentence[1:])
    elif operator == 'implies':
        return (not pl_true(sentence[1], model)) or pl_true(sentence[2], model)
    return False

def generate_truth_table_output(kb, alpha, symbols):
    # Print the exact headers matching your image
    print("P | Q | R || KB    | Query | Row Valid? (KB => Alpha)")
    print("-" * 57)

    all_rows_valid = True

    # Generate all combinations of truth values for P, Q, R
    # (Iterating True first to match the 'T T T' down to 'F F F' order in your table)
    for p_val in [True, False]:
        for q_val in [True, False]:
            for r_val in [True, False]:
                model = {'P': p_val, 'Q': q_val, 'R': r_val}

                # Evaluate formulas
                kb_true = pl_true(kb, model)
                alpha_true = pl_true(alpha, model)

                # Check entailment condition (KB => Alpha)
                if kb_true:
                    row_valid = alpha_true
                else:
                    row_valid = True  # Vacuously true if KB is false

                if not row_valid:
                    all_rows_valid = False

                # Convert to exact short string format (T/F for symbols, True/False for formulas)
                p_str = "T" if p_val else "F"
                q_str = "T" if q_val else "F"
                r_str = "T" if r_val else "F"

                kb_str = "True " if kb_true else "False"
                alpha_str = "True " if alpha_true else "False"
                valid_str = "True" if row_valid else "False"

                print(f"{p_str} | {q_str} | {r_str} || {kb_str} | {alpha_str} | {valid_str}")

    print("\n" + "="*66)
    print(f"Final Result: Does KB entail Query? -> {all_rows_valid}")
    print("="*66)

# Define the scenario from your image:
# KB = P AND Q AND R
# Query = R
kb_expression = ('and', 'P', 'Q', 'R')
query_expression = 'R'
symbols_list = ['P', 'Q', 'R']

# Run code to generate the exact output
generate_truth_table_output(kb_expression, query_expression, symbols_list)
```
