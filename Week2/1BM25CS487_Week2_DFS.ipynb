```python
print("Nandini Kakade")
print("1BM25CS487")
def print_board(board):
    for i in range(0, 9, 3):
        print(board[i:i+3])
    print()

def standard_dfs_solve(start_board, goal_board, max_depth=15):
    goal_tuple = tuple(goal_board)
    visited = set()

    def dfs(board, path):
        if board == goal_tuple:
            return path
        if len(path) >= max_depth:
            return None

        zero_idx = board.index(0)
        r, c = divmod(zero_idx, 3)

        # Standard DFS move order: Up, Down, Left, Right
        moves = {
            "Up": (-1, 0),
            "Down": (1, 0),
            "Left": (0, -1),
            "Right": (0, 1)
        }

        for direction, (dr, dc) in moves.items():
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                neighbor_idx = nr * 3 + nc
                new_board = list(board)
                new_board[zero_idx], new_board[neighbor_idx] = new_board[neighbor_idx], new_board[zero_idx]
                new_board_tuple = tuple(new_board)

                if new_board_tuple not in visited:
                    visited.add(new_board_tuple)
                    result = dfs(new_board_tuple, path + [(direction, new_board_tuple)])
                    if result is not None:
                        return result
                    visited.remove(new_board_tuple)
        return None

    visited.add(tuple(start_board))
    return dfs(tuple(start_board), [])

# Let's run pure DFS on the same states
initial_state = [1, 2, 3, 0, 4, 6, 7, 5, 8]
goal_state = [1, 2, 3, 4, 5, 6, 7, 8, 0]

print("Solving using standard Depth-First Search (DFS):")
std_dfs_solution = standard_dfs_solve(initial_state, goal_state)

if std_dfs_solution:
    print(f"Solved in {len(std_dfs_solution)} steps!\n")
    for step, (move, board) in enumerate(std_dfs_solution, 1):
        print(f"Step {step}: Move 0 {move}")
        print_board(board)
else:
    print("No solution found using standard DFS within the depth limit.")
```
