```python
print("Nandini Kakade")
print("1BM25CS487")

import heapq

# Define the Initial and Goal states as tuples of tuples
INITIAL_STATE = (
    (2, 8, 3),
    (1, 6, 4),
    (7, 0, 5)  # 0 represents the blank space
)

GOAL_STATE = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)

class Node:
    def __init__(self, state, g, parent=None, move=""):
        self.state = state
        self.g = g  # Depth of the node
        self.h = self.calculate_heuristic()  # Number of misplaced tiles
        self.f = self.g + self.h
        self.parent = parent
        self.move = move

    def calculate_heuristic(self):
        """Calculates h(n): number of misplaced tiles excluding the blank space (0)."""
        misplaced = 0
        for i in range(3):
            for j in range(3):
                val = self.state[i][j]
                if val != 0 and val != GOAL_STATE[i][j]:
                    misplaced += 1
        return misplaced

    # Define less-than comparison for the priority queue based on f(n)
    def __lt__(self, other):
        return self.f < other.f

def get_blank_position(state):
    """Finds the coordinates of the blank space (0)."""
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j

def get_neighbors(node):
    """Generates all possible valid moves from the current state."""
    neighbors = []
    r, c = get_blank_position(node.state)

    # Possible moves: (row_change, col_change, direction_name)
    moves = [(-1, 0, "Up"), (1, 0, "Down"), (0, -1, "Left"), (0, 1, "Right")]

    for dr, dc, direction in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            # Convert tuple of tuples to a list of lists to modify the state
            new_state = [list(row) for row in node.state]
            # Swap the blank space with the tile
            new_state[r][c], new_state[nr][nc] = new_state[nr][nc], new_state[r][c]
            # Convert back to tuple of tuples
            new_state_tuple = tuple(tuple(row) for row in new_state)

            neighbors.append(Node(new_state_tuple, node.g + 1, node, direction))

    return neighbors

def print_state(state):
    """Helper function to cleanly display a matrix state."""
    for row in state:
        print(" ".join(str(x) if x != 0 else "_" for x in row))
    print()

def solve_8_puzzle(initial_state):
    """Solves the 8-puzzle using A* Search Algorithm."""
    open_list = []
    closed_set = set()

    start_node = Node(initial_state, g=0)
    heapq.heappush(open_list, start_node)

    while open_list:
        current_node = heapq.heappop(open_list)

        # If the goal state is reached, reconstruct the path
        if current_node.state == GOAL_STATE:
            path = []
            while current_node:
                path.append(current_node)
                current_node = current_node.parent
            return path[::-1]  # Return path from start to goal

        closed_set.add(current_node.state)

        for neighbor in get_neighbors(current_node):
            if neighbor.state in closed_set:
                continue

            # Add to priority queue if it's not already in it with a lower/equal cost
            # (Simplification: heapq handles multiple copies, the lower cost pops first)
            heapq.heappush(open_list, neighbor)

    return None

# --- Execution ---
solution_path = solve_8_puzzle(INITIAL_STATE)

if solution_path:
    print(f"Goal Reached! Total Steps (Depth): {solution_path[-1].g}\n")
    for step, node in enumerate(solution_path):
        if step == 0:
            print("Initial State:")
        else:
            print(f"Step {step} - Move blank {node.move} (g={node.g}, h={node.h}, f={node.f}):")
        print_state(node.state)
else:
    print("No solution found.")
```
